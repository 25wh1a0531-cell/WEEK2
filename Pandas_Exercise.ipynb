```python
import pandas as pd
import numpy as np

print("--- Intermediate Pandas Exercise ---")
print("Complete each task by writing the requested Pandas code.")
print("------------------------------------")
# Task 1.1: Create a Pandas Series from a list
s1_1 = pd.Series([10, 20, 30, 40, 50])

# Task 1.2: Create a 3x4 NumPy array with random integers between 1 and 100
np_array1_2 = np.random.randint(1, 101, size=(3, 4))

# Task 1.3: Convert the NumPy array into a Pandas DataFrame
df1_3 = pd.DataFrame(np_array1_2, columns=['A', 'B', 'C', 'D'])

# Display the results
print("Series:")
print(s1_1)

print("\nNumPy Array:")
print(np_array1_2)

print("\nDataFrame:")
print(df1_3)
```

```python
# Task 2: Create a synthetic sales DataFrame

# Create date range (100 days from 2023-01-01 to 2023-04-10)
dates = pd.date_range(start='2023-01-01', periods=100)

# Create DataFrame
df_sales = pd.DataFrame({
    'Date': dates,
    'Region': np.random.choice(['East', 'West', 'North', 'South'], size=100),
    'Product': np.random.choice(['Laptop', 'Mouse', 'Keyboard', 'Monitor'], size=100),
    'Sales': np.random.randint(100, 1001, size=100).astype(float),      # Convert to float to allow NaN
    'Quantity': np.random.randint(1, 11, size=100).astype(float)         # Convert to float to allow NaN
})

# Introduce 5 random NaN values in the 'Sales' column
sales_nan_indices = np.random.choice(df_sales.index, size=5, replace=False)
df_sales.loc[sales_nan_indices, 'Sales'] = np.nan

# Introduce 3 random NaN values in the 'Quantity' column
quantity_nan_indices = np.random.choice(df_sales.index, size=3, replace=False)
df_sales.loc[quantity_nan_indices, 'Quantity'] = np.nan

# Display the first 10 rows
print(df_sales.head(10))
```

```python
# Task 3.1: Select only the 'Region' and 'Sales' columns
df3_1 = df_sales[['Region', 'Sales']]

# Task 3.2: Select rows where 'Product' is 'Laptop'
df3_2 = df_sales[df_sales['Product'] == 'Laptop']

# Task 3.3: Select rows where 'Region' is 'East' AND 'Sales' are greater than 500
df3_3 = df_sales[(df_sales['Region'] == 'East') & (df_sales['Sales'] > 500)]

# Task 3.4: Select the 'Sales' value for the first entry where 'Product' is 'Monitor'
val3_4 = df_sales.loc[df_sales['Product'] == 'Monitor', 'Sales'].iloc[0]

# Display the results
print("Task 3.1:")
print(df3_1.head())

print("\nTask 3.2:")
print(df3_2.head())

print("\nTask 3.3:")
print(df3_3.head())

print("\nTask 3.4:")
print(val3_4)
```

```python
# Task 4.1: Check the number of missing values in each column
missing_counts4_1 = df_sales.isnull().sum()

# Task 4.2: Fill missing 'Sales' values with the mean of the 'Sales' column
df4_2_filled = df_sales.copy()
df4_2_filled['Sales'] = df4_2_filled['Sales'].fillna(df4_2_filled['Sales'].mean())

# Task 4.3: Drop rows that have any missing values
df4_3_dropped = df_sales.dropna()

# Display the results
print("Task 4.1: Missing values in each column")
print(missing_counts4_1)

print("\nTask 4.2: DataFrame with missing 'Sales' values filled")
print(df4_2_filled.head())

print("\nTask 4.3: DataFrame after dropping rows with missing values")
print(df4_3_dropped.head())
```

```python
# Task 5.1: Create a new column 'Total_Revenue'
df_sales['Total_Revenue'] = df_sales['Sales'] * df_sales['Quantity']

# Task 5.2: Sort by 'Date' (ascending) and then by 'Sales' (descending)
df5_2_sorted = df_sales.sort_values(by=['Date', 'Sales'], ascending=[True, False])

# Task 5.3: Group by 'Region' and calculate the sum of 'Sales' and 'Quantity'
df5_3_grouped_region = df_sales.groupby('Region')[['Sales', 'Quantity']].sum()

# Task 5.4: Group by 'Product' and calculate the average 'Sales'
df5_4_grouped_product = df_sales.groupby('Product')['Sales'].mean()

# Display the results
print("Task 5.1: DataFrame with Total_Revenue")
print(df_sales.head())

print("\nTask 5.2: Sorted DataFrame")
print(df5_2_sorted.head())

print("\nTask 5.3: Total Sales and Quantity by Region")
print(df5_3_grouped_region)

print("\nTask 5.4: Average Sales by Product")
print(df5_4_grouped_product)
```
